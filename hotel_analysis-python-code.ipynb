EXECUTIVE KPI's

```python
TOTAL BOOKING
```

```python
total_bookings = len(df)
```

```python
average_stay = df["total_nights"].mean()
```

```python
Cancellation by market segment: Which customer acquisition channels bring higher cancellation risk
```

```python
segment_cancellation = (
    df.groupby("market_segment")["is_canceled"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

segment_cancellation
```

```python
segment_cancellation.plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Cancellation Rate by Market Segment")
plt.xlabel("Market Segment")
plt.ylabel("Cancellation Rate (%)")
plt.xticks(rotation=45)
plt.show()
```

```python
DISTRIBUTION CHANNEL ANALYSIS
```

```python
df["distribution_channel"].value_counts()
```

```python
channel_cancellation = (
    df.groupby("distribution_channel")["is_canceled"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

channel_cancellation
```

```python
channel_cancellation.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Cancellation Rate by Distribution Channel")
plt.xlabel("Distribution Channel")
plt.ylabel("Cancellation Rate (%)")
plt.show()
```

```python
DEPOSIT TYPE ANALYSIS Does the deposit policy influence cancellation behavior
```

```python
deposit_cancellation = (
    df.groupby("deposit_type")["is_canceled"]
    .mean()
    .mul(100)
)

deposit_cancellation
```

```python
sns.barplot(
    data=deposit_cancellation.reset_index(),
    x="deposit_type",
    y="is_canceled"
)

plt.title("Cancellation Rate by Deposit Type")
plt.xlabel("Deposit Type")
plt.ylabel("Cancellation Rate (%)")
plt.show()
```

```python
resort_hotel=resort_hotel.groupby("reservation_status_date")[["adr"]].mean()
city_hotel=city_hotel.groupby("reservation_status_date")[["adr"]].mean()
```

```python
plt.figure(figsize=(20,8))
plt.title("average daily rate in city and resort hotel")
plt.plot(resort_hotel.index,resort_hotel["adr"],label="resort hotel")
plt.legend(fontsize=30)
plt.show()
```

```python
plt.figure(figsize=(20,8))
plt.title("average daily rate in city and resort hotel")
plt.plot(city_hotel.index,city_hotel["adr"],label="city hotel")
plt.legend(fontsize=30)
plt.show()
```

```python
plt.figure(figsize=(20,8))
plt.title("average daily rate in city and resort hotel")
plt.plot(resort_hotel.index,resort_hotel["adr"],label="resort hotel")
plt.plot(city_hotel.index,city_hotel["adr"],label="city hotel")
plt.legend(fontsize=30)
plt.show()
```

```python
df["month"]=df["reservation_status_date"].dt.month
plt.figure(figsize=(16,8))
ax1=sns.countplot(x="month",hue="is_canceled",data=df,palette="bright")
legend_labels,_=ax1.get_legend_handles_labels()
ax1.legend(bbox_to_anchor=(1,1))
plt.ylabel("number of resrvations")
plt.legend("not cancelled","cancelled")
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.title("adr per month",fontsize=30)
sns.barplot(x="month" ,y="adr",data=df[df["is_canceled"]==1].groupby("month")["adr"].sum().reset_index())

plt.legend(fontsize=20)
plt.show()
```
